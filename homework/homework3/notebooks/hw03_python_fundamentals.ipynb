```python
import os
import sys
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --------------------------------------------------------------------------
# Path setup: locate the project folder (the folder *containing* this
# notebook) so the data files load correctly regardless of the kernel's cwd.
# --------------------------------------------------------------------------
def _here() -> Path:
    # VS Code exposes the notebook's own path through this global variable.
    vs_path = globals().get("__vsc_ipynb_file__")
    if vs_path:
        return Path(vs_path).resolve().parent
    # Jupyter classic / other kernels: fall back to the current directory.
    return Path.cwd().resolve()


NB_DIR = _here()                     # .../homework3/notebooks
PROJECT_DIR = NB_DIR.parent          # .../homework3
RAW_DIR = PROJECT_DIR / "data" / "raw"
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"

# Make src/ (a sibling of notebooks/) importable.
sys.path.insert(0, str(PROJECT_DIR))

plt.style.use("ggplot")
```

## Step 1 — NumPy Operations

Create an array, apply elementwise operations, then compare a plain Python
loop with a vectorized NumPy expression.


```python
# --- Elementwise operations on a NumPy array ---
arr = np.arange(1, 11)                       # 1, 2, ..., 10
print("array        :", arr)
print("arr ** 2     :", arr ** 2)            # elementwise square
print("arr + 5      :", arr + 5)
print("np.sqrt(arr) :", np.sqrt(arr).round(3))

# --- Loop vs vectorized execution ---
big = np.arange(1_000_000)

print("\nSquare every element of 1,000,000 numbers:")
print("Python loop:")
%timeit [x * 2 for x in big]
print("NumPy vectorized:")
%timeit big * 2
```

## Step 2 — Dataset Loading

Load `data/raw/starter_data.csv` with pandas and inspect it using `.info()`
and `.head()`.


```python
csv_path = RAW_DIR / "starter_data.csv"
df = pd.read_csv(csv_path)

print("Shape:", df.shape)
print("\n--- df.head() ---")
display(df.head())

print("\n--- df.info() ---")
df.info()

print("\n--- column dtypes ---")
print(df.dtypes)
```

## Step 3 — Summary Statistics

- `.describe()` summarises the numeric columns.
- `.groupby("category")` aggregates values per category.


```python
# --- Descriptive statistics for the numeric columns ---
print("--- df.describe() ---")
display(df.describe())

# --- Group-by aggregation: mean value per category ---
summary = df.groupby("category").mean(numeric_only=True).reset_index()
print("--- Mean value by category ---")
display(summary)
```

## Step 4 — Save Outputs

Save the summary table to `data/processed/summary.csv` (plus a JSON copy).
**Bonus:** create and save a basic histogram.


```python
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Required: save the summary stats to CSV (and a JSON copy).
summary.to_csv(PROCESSED_DIR / "summary.csv", index=False)
summary.to_json(PROCESSED_DIR / "summary.json", orient="records", indent=2)
print("Saved:", PROCESSED_DIR / "summary.csv")
print("Saved:", PROCESSED_DIR / "summary.json")

# Bonus: a basic histogram of the numeric column.
plt.figure(figsize=(7, 4))
df["value"].plot(kind="hist", bins=8, edgecolor="white")
plt.title("Distribution of value")
plt.xlabel("value")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(PROCESSED_DIR / "histogram.png", dpi=120)
plt.show()
```
